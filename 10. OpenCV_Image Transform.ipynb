```python
import cv2
import numpy as np
```

```python
src = cv2.imread("./Data/example.png")

# 회전 중심 설정
(h, w) = src.shape[:2]
center = (100, 100)

# 회전을 위한 매트릭스 생성
M = cv2.getRotationMatrix2D(center, 45, 1.0)

# OpenCV를 사용하여 이미지 회전
img_rotated = cv2.warpAffine(src, M, (w, h))

cv2.imshow("src", src)
cv2.imshow("img_rotated", img_rotated)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
flipVertical = cv2.flip(src, 0) # 상하 반전(FlipCode = 0)
flipHorizontal = cv2.flip(src, 1) # 좌우 반전(FlipCode > 0)
flipBoth = cv2.flip(src, -1) # 상하좌우 반전(FlipCode < 0)

cv2.imshow("src", src)
cv2.imshow("flipVertical", flipVertical)
cv2.imshow("flipHorizontal", flipHorizontal)
cv2.imshow("flipBoth", flipBoth)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 새로운 크기 설정
new_width = 300
new_height = 200
new_size = (new_width, new_height)

# 이미지 크기 변경
img_resized = cv2.resize(src, new_size)

cv2.imshow("src", src)
cv2.imshow("Resized Image", img_resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 상대적인 크기 조절을 위한 배율 설정 (ex. 너비를 0.5배, 높이를 0.5배로 조절)
fx = 1.5
fy = 1.5

# 영역 보간법
img_resized_area = cv2.resize(src, None, fx=fx, fy=fy, interpolation=cv2.INTER_AREA)
# 최근접 이웃 보간법
img_resized_nearest = cv2.resize(src, None, fx=fx, fy=fy, interpolation=cv2.INTER_NEAREST)
# 선형 보간법
img_resized_linear = cv2.resize(src, None, fx=fx, fy=fy, interpolation=cv2.INTER_LINEAR)
# 3차 보간법
img_resized_cubic = cv2.resize(src, None, fx=fx, fy=fy, interpolation=cv2.INTER_CUBIC)
# 란초스 보간법
img_resized_lanczos4 = cv2.resize(src, None, fx=fx, fy=fy, interpolation=cv2.INTER_LANCZOS4)

cv2.imshow("Resized Image A", img_resized_area)
cv2.imshow("Resized Image N", img_resized_nearest)
cv2.imshow("Resized Image L", img_resized_linear)
cv2.imshow("Resized Image C", img_resized_cubic)
cv2.imshow("Resized Image Lanczos4", img_resized_lanczos4)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
canvas = np.zeros(shape=(800, 800, 3), dtype=np.uint8)

# 사다리꼴 예제 생성
pts1 = np.array([[320, 15], [700, 215], [530, 780], [85, 610]])

cv2.polylines(img=canvas,
              pts=[pts1],
              isClosed=True,
              color=(0, 255, 0),
              thickness=3)

cv2.imshow("canvas", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 사다리꼴 4개 꼭짓점 좌표 설정 (실제 이미지에 맞게 조정해야함)
pts1 = np.float32(pts1)
# 변환될 사각형의 4개 꼭짓점 좌표 설정 (현재는 직사각형)
pts2 = np.float32([[100, 100], [600, 100], [600, 400], [100, 400]])

# 투시 변환 매트릭스 계산
M = cv2.getPerspectiveTransform(pts1, pts2)

# 투시 변환 적용
img_transformed = cv2.warpPerspective(canvas, M, (1000, 600))

cv2.imshow("canvas", canvas)
cv2.imshow("img_transformed", img_transformed)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
src = cv2.imread("./Data/ticket.png")

points = []
click_count = 0

def select_points(event, x, y, flags, param):
    global points, click_count
    
    # 왼쪽 마우스 버튼이 클릭될 때 좌표 저장
    if event == cv2.EVENT_LBUTTONDOWN:
        points.append((x, y))
        click_count += 1
        cv2.imshow("Image", src)

# 'Image' 윈도우에서 마우스 콜백 함수 설정
cv2.namedWindow("Image")
cv2.setMouseCallback("Image", select_points)

# 사용자가 4개의 점을 클릭할 때까지 기다림
while click_count < 4:
    cv2.imshow("Image", src)
    cv2.waitKey(1)

cv2.destroyAllWindows()

# 선택한 점을 사용하여 사다리꼴의 꼭짓점 설정
pts1 = np.float32(points)

# 변환될 사각형의 4개 꼭짓점 좌표 설정
pts2 = np.float32([[0, 0], [0, 300], [500, 300], [500, 0]])

# 투시 변환 매트릭스 계산
M = cv2.getPerspectiveTransform(pts1, pts2)

# 투시 변환 적용
img_transformed = cv2.warpPerspective(src, M, (500, 300))

cv2.imshow("src", src)
cv2.imshow("img_transformed", img_transformed)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
