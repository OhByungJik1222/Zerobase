```python
# 임의의 데이터로 Linear Regression 이해
import pandas as pd

data = {'x': [1., 2., 3., 4., 5.],
        'y': [1., 3., 4., 6., 5.]}
df = pd.DataFrame(data)
df
```

```python
# 데이터 시각화
import matplotlib.pyplot as plt

plt.scatter(df['x'], df['y'])
plt.grid()
plt.show()
```

```python
# Linear Regression 학습
from sklearn.linear_model import LinearRegression

x = df[['x']]
y = df['y']

lr = LinearRegression()
lr.fit(x, y)
```

```python
# Root Mean Absolute Error
import numpy as np
from sklearn.metrics import mean_absolute_error

pred = lr.predict(x)
rmae = np.sqrt(mean_absolute_error(y, pred))
rmae
```

```python
# 실제 vs 예측값 비교
# 이때 예측이 완벽할 때는 y = x 함수에 정확히 매칭됨
plt.figure(figsize=(8, 8))

plt.scatter(y, pred)
plt.plot([0, 6], [0, 6], 'r')

plt.title('Real vs. Predicted')
plt.xlabel('Actual')
plt.ylabel('Predicted')

plt.grid()
plt.show()
```
