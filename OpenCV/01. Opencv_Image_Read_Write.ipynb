```python
video_path = "./Data/Sample Video.avi"

cap = cv.VideoCapture(video_path)

frame_count = int(cap.get(cv.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv.CAP_PROP_FPS)
width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))

print("동영상 정보 :" + str(width) + "x" + str(height) + ", " + str(frame_count) + " 프레임, " + str(fps) + "FPS")
```

```python
save_folder = "./frames"
os.makedirs(save_folder, exist_ok=True)

padding_length = len(str(frame_count))

cap = cv.VideoCapture(video_path)

frame_index = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break

    padded_frame_idx = str(frame_index).zfill(padding_length)
    save_path = save_folder + "/frame_" + padded_frame_idx + ".jpg"
    cv.imwrite(save_path, frame)

    frame_index += 1

cap.release()

print("프레임 저장 완료")
```

```python
output_video_path = "output_video.avi"

# 동영상의 해상도와 FPS 설정
frame_width = 640
frame_height = 360
fps = 30

# VideoWriter 객체 생성
fourcc = cv.VideoWriter_fourcc(*"XVID")
out = cv.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# 프레임 파일 목록 가져오기
frame_files = os.listdir("./Data/frames")
frame_files.sort()

# 각 프레임을 읽어서 동영상 파일에 추가
for file in frame_files:
    frame_path = os.path.join("./Data/frames", file)
    frame = cv.imread(frame_path)

    if frame is not None:
        # 프레임 크기 조정
        frame = cv.resize(frame, (frame_width, frame_height))
        out.write(frame)

# 작업 완료 후 자원 해제
out.release()

print("동영상 저장 완료")
```

```python
frame_files = os.listdir("./Data/frames")

# 2초 분량의 프레임만 사용
frames = []
for file in frame_files[:60]: # 30fps 기준 2초는 60frame
    frame_path = os.path.join("./Data/frames", file)
    frame = cv.imread(frame_path)

    if frame is not None:
        frames.append(frame)

output_video_path = "boomerang_video.avi"
fourcc = cv.VideoWriter_fourcc(*"XVID")
out = cv.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# 5번 반복
for _ in range(5):
    # 순방향 재생
    for frame in frames:
        out.write(frame)
    # 역방향 재생
    for frame in reversed(frames[:]):
        out.write(frame)

# 자원 해제
out.release()

print("부메랑 영상 저장 완료!")
```

```python
import time

# 카메라 설정
cap = cv.VideoCapture(0) # 0은 첫번째 카메라를 의미

# 영상 저장을 위한 VideoWriter 설정
width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH)) # XVID 코덱 사용
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))

fourcc = cv.VideoWriter_fourcc(*"XVID")
out = cv.VideoWriter("cam_test.avi", fourcc, 20.0, (width, height))

time.sleep(5)

while True:
    # 카메라로부터 프레임 읽기
    ret, frame = cap.read()
    if not ret:
        break

    # 화면에 프레임 표시
    cv.imshow("Camera", frame)

    out.write(frame)

    # 'q' 키를 누르면 루프 탈출
    if cv.waitKey(1) & 0xFF == ord("q"):
        break

# 자원 해제
cap.release()
out.release()
cv.destroyAllWindows()
```
