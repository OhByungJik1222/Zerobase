```python
# 이미지 불러오기
img_path = "./example.png"
img = cv2.imread(img_path)

img_R = img.copy()

# 다른 R 제외 채널 값 제거
img_R[:,:,0] = 0 # B
img_R[:,:,1] = 0 # G

cv2.imshow("original", img)
cv2.imshow("Red Channel", img_R)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 이미지를 반전시키기 위해 RGB 값을 255에서 뺀다
inverted_img = 255 - img

cv2.imshow("original", img) # Original Image
cv2.imshow("Inverted imgae", inverted_img) # Inverted Image

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
add_value = 255

# unit8 데이터 타입은 0 ~ 255까지만 저장
# 따라서 255를 초과하는 값은 256로 나눈 나머지로 저장됨(Modular)
added_img = src + add_value

cv2.imshow("src", src)
cv2.imshow("added_img", added_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
add_value = 255

# 밝기 조절을 위한 뎃셈
# 모두 255가 되어 그냥 하얀색 이미지
added_img = cv2.add(src, np.full(src.shape, add_value, dtype=np.uint8))

cv2.imshow("src", src)
cv2.imshow("added_img", added_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
add_value = 100

# 밝기 올리기
added_img = cv2.add(src, np.full(src.shape, add_value, dtype=np.uint8))

cv2.imshow("src", src)
cv2.imshow("added_img", added_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
sub_value = 100

# 밝기 낮추기
sub_img = cv2.subtract(src, np.full(src.shape, sub_value, dtype=np.uint8))

cv2.imshow("src", src)
cv2.imshow("added_img", sub_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# RGB 이미지를 그레이 그케일로 변환
gray_img = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)

cv2.imshow("src", src)
cv2.imshow("gray_img", gray_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# RGB 이미지를 HSV로 변환
hsv_img = cv2.cvtColor(src, cv2.COLOR_BGR2HSV)

cv2.imshow("src", src)
cv2.imshow("hsv_img", hsv_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
hsv_img = cv2.cvtColor(src, cv2.COLOR_BGR2HSV)

# 빨간색 영역만 골라내기
hsv_red = cv2.inRange(hsv_img, (0, 0, 0), (30, 0, 0))
# 파란색 영역만 골라내기
hsv_blue = cv2.inRange(hsv_img, (110, 0, 0), (130, 255, 255))

cv2.imshow("src", src)
cv2.imshow("hsv_red", hsv_red)
cv2.imshow("hsv_blue", hsv_blue)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
