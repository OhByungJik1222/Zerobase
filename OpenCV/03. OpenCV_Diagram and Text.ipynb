```python
import cv2
import numpy as np

# 빈 캔버스 생성 (검은색 배경)
canvas = np.zeros(shape=(500, 500, 3), dtype=np.uint8)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 사각형의 시작점 좌표와 끝점 좌표
start_point = (150, 150)
end_point = (350, 350)

# 사각형 그리기 (내부가 빈 사각형)
cv2.rectangle(img = canvas,
              pt1 = start_point,
              pt2 = end_point,
              color = (0, 255, 255),
              thickness = 3)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 사각형의 시작점 좌표와 끝점 좌표
start_point = (150, 353)
end_point = (350, 450)

# 사각형 그리기 (내부가 차있는 사각형)
cv2.rectangle(img = canvas,
              pt1 = start_point,
              pt2 = end_point,
              color = (42, 42, 165),
              thickness = -1)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 원의 중심 좌표
center_coordinates = (220, 220)

# 원 그리기
cv2.circle(img = canvas,
           center = (220, 220),
           radius = 30,
           color = (255, 255, 255),
           thickness = 0)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 타원의 중심 좌표
center_coordinates = (280, 250)

# 타원 그리기
cv2.ellipse(img = canvas,
            center = center_coordinates,
            axes = (30, 10), # 타원의 축 길이
            angle = 90,
            startAngle = 0,
            endAngle = 360,
            color = (0, 255, 255),
            thickness = 0)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 타원의 중심 좌표
center_coordinates = (300, 300)

# 타원 그리기
cv2.ellipse(img = canvas,
            center = center_coordinates,
            axes = (30, 10), # 타원의 축 길이
            angle = 0,
            startAngle = 0,
            endAngle = 180, # 반만 그리기
            color = (0, 0, 255),
            thickness = 0)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
canvas = np.zeros(shape=(500, 500, 3), dtype=np.uint8)

# 꼭짓점 좌표
points = np.array([[250, 250], [150, 350], [350, 350]])

# 다각형 그리기 (삼각형)
cv2.polylines(img = canvas,
              pts = [points], # 꼭짓점
              isClosed = True, # 처음 좌표와 마지막 좌표를 이어줌
              color = (0, 0, 255),
              thickness = 3)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
canvas = np.zeros(shape=(500, 500, 3), dtype=np.uint8)

# 꼭짓점 좌표
points = np.array([[250, 150], [150, 350], [300, 150], [350, 350]])

# 다각형 그리기
cv2.polylines(img = canvas,
              pts = [points],
              isClosed = True,
              color = (0, 0, 255),
              thickness = 3)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 선 그리기
cv2.line(img = canvas,
         pt1 = (150, 350),
         pt2 = (350, 350),
         color = (255, 0, 0),
         thickness = 2)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 점 그리기
cv2.circle(img = canvas,
           center = (250, 250),
           radius = 10,
           color = (255, 0, 255),
           thickness = -1)

cv2.imshow("Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread("./Data/example.png")

# 텍스트 삽입
text = "Nabi"
font = cv2.FONT_HERSHEY_SIMPLEX
position = (50, 50) # 텍스트가 시작될 좌표
font_scale = 1 # 폰트 크기
color = (0, 255, 0) # 녹색 텍스트
thickness = 2 # 텍스트 두께

cv2.putText(img, text, position, font, font_scale, color, thickness)

cv2.imshow("Img with Text", img)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
from PIL import Image, ImageDraw, ImageFont

img = cv2.imread("./Data/example.png")

# OpenCV 이미지를 PIL 이미지로 변환
img_pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

# PIL을 이용하여 텍스트 그리기
draw = ImageDraw.Draw(img_pil)

# 원하는 폰트 파일 경롤로 변경
font = ImageFont.truetype("C:/Windows/Fonts/malgun.ttf", 40)

draw.text((50, 50),
          "나비",
          font=font,
          fill = (0, 255, 0))

# PIL 이미지를 다시 OpenCV 이미지로 변환
img = cv2.cvtColor(np.array(img_pil), cv2.COLOR_RGB2BGR)

cv2.imshow("Img with Korean", img)
cv2.waitKey(0)

cv2.destroyAllWindows()
```
