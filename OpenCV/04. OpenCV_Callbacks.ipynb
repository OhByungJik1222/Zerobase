```python
import cv2
import numpy as np
```

```python
# 마우스 콜백 함수 (좌표 출력하기)
def click_event(event, x, y, flags, params):
    if event == cv2.EVENT_LBUTTONDOWN:
        print("Coordinates of Point:", x, y)

src = cv2.imread("./Data/example.png")
cv2.imshow("src", src)

# 마우스 콜백 함수 설정
cv2.setMouseCallback("src", click_event)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 마우스 콜백 함수 (원 그리기)
def mouse_draw_circle(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        cv2.circle(src, (x, y), 5, (255, 0, 0), -1)
        cv2.imshow("src", src)

src = cv2.imread("./Data/example.png")
cv2.imshow("src", src)

# 마우스 콜백 함수 설정
cv2.setMouseCallback("src", mouse_draw_circle)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 마우스 콜백 함수 (선 그리기)
def callback(event, x, y, flags, param):
    global canvas, img_temp, drawing, start_point

    if event == cv2.EVENT_LBUTTONDOWN: # 왼쪽 마우스를 누름
        img_temp = canvas.copy()
        drawing = True
        start_point = (x, y)

    elif event == cv2.EVENT_MOUSEMOVE: # 마우스를 움직임
        canvas = img_temp.copy()
        if drawing:
            cv2.line(canvas, start_point, (x, y), (255, 255, 255), 5)

    elif event == cv2.EVENT_LBUTTONUP: # 왼쪽 마우스를 뗌
        cv2.line(canvas, start_point, (x, y), (255, 255, 255), 5)
        drawing = False
        img_temp = canvas.copy()


img_temp = None
drawing = False
start_point = None

canvas = np.zeros((512, 512, 3), np.uint8)

cv2.imshow("canvas", canvas)
cv2.setMouseCallback("canvas", callback)

while(True):

    cv2.imshow("canvas", canvas)
    
    key = cv2.waitKey(1)
    if key == 27: # ESC 입력 시 루프 탈출
        break
        
cv2.destroyAllWindows()
```

```python
# 마우스로 선 그리기 & CTRL 키로 색상 바꾸기
def callback(event, x, y, flags, param):
    global canvas, img_temp, drawing, start_point

    if event == cv2.EVENT_LBUTTONDOWN:
        img_temp = canvas.copy()
        drawing = True
        start_point = (x, y)

    elif event == cv2.EVENT_MOUSEMOVE:
        canvas = img_temp.copy()
        if drawing:
            color = (255, 255, 255) # 기본 색상: 흰색

            # CTRL 키를 누르고 잇을 경우 빨간색으로 그림
            if flags & cv2.EVENT_FLAG_CTRLKEY:
                color = (0, 0, 255)
            
            cv2.line(canvas, start_point, (x, y), color, 5)

    elif event == cv2.EVENT_LBUTTONUP:
        color = (255, 255, 255)

        if flags & cv2.EVENT_FLAG_CTRLKEY:
            color = (0, 0, 255)
        
        cv2.line(canvas, start_point, (x, y), color, 5)
        drawing = False
        img_temp = canvas.copy()


img_temp = None
drawing = False
start_point = None

canvas = np.zeros((512, 512, 3), np.uint8)

cv2.imshow("canvas", canvas)
cv2.setMouseCallback("canvas", callback)

while(True):

    cv2.imshow("canvas", canvas)
    
    key = cv2.waitKey(1)
    if key == 27:
        break
        
cv2.destroyAllWindows()
```

```python
# 트랙바로 색상 찾기
img = np.zeros((300, 600, 3), np.uint8)

# 트랙바 콜백 함수
def nothing(x):
    pass

cv2.namedWindow("image")

# RGB 트랙바 생성
cv2.createTrackbar('R', 'image', 0, 255, nothing)
cv2.createTrackbar('G', 'image', 0, 255, nothing)
cv2.createTrackbar('B', 'image', 0, 255, nothing)

while True:
    cv2.imshow('image', img)

    key = cv2.waitKey(1)
    if key == 27: # ESC 누르면 루프 탈출
        break

    # 트랙바 위치 읽기
    r = cv2.getTrackbarPos('R', 'image')
    g = cv2.getTrackbarPos('G', 'image')
    b = cv2.getTrackbarPos('B', 'image')

    # 읽은 트랙바 위치에 따라 이미지 색상 변경
    img[:] = [b, g, r]

cv2.destroyAllWindows()
```

```python
# HSV에 적용
img = np.zeros((300, 600, 3), np.uint8)

def nothing(x):
    pass

cv2.namedWindow("HSV_Image")

cv2.createTrackbar('H', 'HSV_Image', 0, 179, nothing)
cv2.createTrackbar('S', 'HSV_Image', 0, 255, nothing)
cv2.createTrackbar('V', 'HSV_Image', 0, 255, nothing)

while True:
    cv2.imshow('HSV_Image', img)

    key = cv2.waitKey(1)
    if key == 27:
        break

    h = cv2.getTrackbarPos('H', 'HSV_Image')
    s = cv2.getTrackbarPos('S', 'HSV_Image')
    v = cv2.getTrackbarPos('V', 'HSV_Image')

    hsv_color = np.uint8([[[h, s, v]]])
    rgb_color = cv2.cvtColor(hsv_color, cv2.COLOR_HSV2BGR)
    
    img[:] = rgb_color

cv2.destroyAllWindows()
```
