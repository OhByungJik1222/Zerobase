```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
plane_day = cv2.imread("./Data/plane_day.png")

# 흑백 이미지 히스토그램
gray_plane_day = cv2.cvtColor(plane_day, cv2.COLOR_BGR2GRAY)
hist_day = cv2.calcHist([gray_plane_day], [0], None, [256], [0, 256])

# 흑백 이미지 히스토그램 그래프
plt.plot(hist_day)
plt.xlim([0, 256])
plt.title("GrayScale hist")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")

plt.show()

cv2.imshow("plane_day", plane_day)
cv2.imshow("gray_plane_day", gray_plane_day)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plane_night = cv2.imread("./Data/plane_night.png")

gray_plane_night = cv2.cvtColor(plane_night, cv2.COLOR_BGR2GRAY)
hist_night = cv2.calcHist([gray_plane_night], [0], None, [256], [0, 256])

plt.plot(hist_night)
plt.xlim([0, 256])
plt.title("GrayScale hist (night)")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")

plt.show()

cv2.imshow("plane_night", plane_night)
cv2.imshow("gray_plane_night", gray_plane_night)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plane_day = cv2.imread("./Data/plane_day.png")

# 컬러 이미지 히스토그램 (BGR)
color = ('b', 'g', 'r')
for i, col in enumerate(color):
    color_hist_day = cv2.calcHist([plane_day], [i], None, [256], [0, 256])
    plt.plot(color_hist_day, color = col)


plt.xlim([0, 256])
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")

plt.show()

cv2.imshow("plane_day", plane_day)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 이미지 히스토그램 평활화
# - 이미지의 밝기 히스토그램을 조절하여 전체적인 대비를 개선
# - 이미지의 대비를 향상시키는 데 사용
plane_night = cv2.imread("./Data/plane_night.png")

gray_plane_night = cv2.cvtColor(plane_night, cv2.COLOR_BGR2GRAY)
hist_night = cv2.calcHist([gray_plane_night], [0], None, [256], [0, 256])

# 히스토그램 평활화
eq_img = cv2.equalizeHist(gray_plane_night)
hist_eq = cv2.calcHist([eq_img], [0], None, [256], [0, 256])

plt.plot(hist_night)
plt.xlim([0, 256])
plt.title("GrayScale hist (night)")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")
plt.show()

plt.plot(hist_eq)
plt.xlim([0, 256])
plt.title("GrayScale hist (eq)")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")
plt.show()

cv2.imshow("plane_night", plane_night)
cv2.imshow("gray_plane_night", gray_plane_night)
cv2.imshow("eq_img", eq_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 테스트용 이미지 생성 (일부 밝고 일부 어두운 영역 포함)
image = np.zeros((500, 500), dtype=np.uint8)

image[0:200, :] = 30 # 상단은 어두운 영역
image[200:300, :] = 120 # 중간은 중간 밝기 영역
image[300:500, :] = 220 # 하단은 밝은 영역

# 원본 이미지의 히스토그램 계산
hist_origin = cv2.calcHist([image], [0], None, [256], [0, 256])

# 히스토그램 평활화
img_eq = cv2.equalizeHist(image)
hist_eq = cv2.calcHist([img_eq], [0], None, [256], [0, 256])

# 히스토그램 스트레칭 (정규화)
img_norm = cv2.normalize(image, None, 0, 255, cv2.NORM_MINMAX)
hist_norm = cv2.calcHist([img_norm], [0], None, [256], [0, 256])


plt.plot(hist_origin)
# plt.xlim([0, 256])
plt.title("hist_origin")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")
plt.show()

plt.plot(hist_eq)
# plt.xlim([0, 256])
plt.title("hist_eq")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")
plt.show()

plt.plot(hist_norm)
# plt.xlim([0, 256])
plt.title("hist_norm")
plt.xlabel("Bins")
plt.ylabel("Number of Pixels")
plt.show()

cv2.imshow("image", image)
cv2.imshow("img_eq", img_eq)
cv2.imshow("img_norm", img_norm)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
