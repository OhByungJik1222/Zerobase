```python
import cv2
import numpy as np
```

```python
image = np.full((200, 200, 3), 255, np.uint8)

cv2.rectangle(image, (20, 20), (180, 180), (0, 0, 0), -1)

white_noise = np.random.randint(0, 256, (200, 200, 3), dtype=np.uint8)
img_noisy = cv2.addWeighted(image, 0.5, white_noise, 0.5, 0)

cv2.imshow("img_noisy", img_noisy)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blur = cv2.blur(img_noisy, (3, 3))
blur2 = cv2.blur(blur, (3, 3))

cv2.imshow("img_noisy", img_noisy)
cv2.imshow("blur", blur)
cv2.imshow("blur2", blur2)


cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blur = cv2.blur(img_noisy, (3, 3)) # 일반 블러
gaussian_blur = cv2.GaussianBlur(img_noisy, (3, 3), 0) # 가우시안 블러
median_blur = cv2.medianBlur(img_noisy, 3) # 미디언 블러
bilateral_filter = cv2.bilateralFilter(img_noisy, -1, 75, 75) # 양방향 블러

cv2.imshow("img_noisy", img_noisy)
cv2.imshow("blur", blur)
cv2.imshow("gaussian_blur", gaussian_blur)
cv2.imshow("median_blur", median_blur)
cv2.imshow("bilateral_filter", bilateral_filter)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
