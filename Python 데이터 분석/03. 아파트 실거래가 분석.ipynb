## **correlation_coefficient**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
family = 'Malgun Gothic'

plt.rc('font', family=family)
```

```python
sample = 100

x = np.random.rand(sample)
y = x + np.random.rand(sample) * 0.5
```

```python
df = pd.DataFrame([x, y]).T
df.head(10)
```

```python
def draw_graph(x, y):
    plt.scatter(x, y)
    plt.scatter(np.mean(x), np.mean(y), c='red')
    plt.plot([np.mean(x), np.mean(x)], [min(y), max(y)], marker='o', color='red')
    plt.plot([min(x), max(x)], [np.mean(y), np.mean(y)], marker='o', color='red')
    plt.grid(True)

    plt.show()
```

```python
draw_graph(x, y)
```

```python
# 양의 상관관계를 가진다.
df.corr()
```

## **scatterplot**

```python
df = pd.read_csv('.\data\서울특별시\apartment_df.csv')
df.head(10)
```

```python
location = pd.read_csv('.\동별좌표(구).csv')
location.head(10)
```

```python
dong_data = location[location['동'] == '한남동']
dong_data.head(10)
```

```python
dong_data.iloc[0]
```

```python
def search_location(dong):
    try:
        dong_data = location[location['동'] == dong].iloc[0]
        return (dong_data['loc'], dong_data['lat'])
    except:
        print(dong)
        return (np.nan, np.nan)
```

```python
len(np.unique(df['dong']))
```

```python
price_by_dong = df.groupby('dong')['transaction_real_price'].mean().reset_index()
price_by_dong.head(20)
```

```python
count_by_dong = df.groupby('dong')['transaction_real_price'].count().reset_index()
count_by_dong.head(20)
```

```python
merged_df = pd.merge(price_by_dong, count_by_dong, on='dong')
merged_df.columns = ['dong', 'mean_price', 'count']
merged_df.head(20)
```

```python
merged_df[['loc', 'lat']] = merged_df['dong'].apply(search_location).apply(pd.Series)
merged_df.head(20)
```

```python
sns.scatterplot(data=merged_df,
                x='lat',
                y='loc',
                hue='dong',
                size='mean_price',
                sizes=(10, 200),
                legend=False)
plt.show()
```

```python
fig = px.scatter(merged_df,
                 x='lat',
                 y='loc',
                 color='dong',
                 size='mean_price')
fig.show()
```

```python
fig = px.scatter(merged_df,
                 x='lat',
                 y='loc',
                 color='dong',
                 size='count')
fig.show()
```
