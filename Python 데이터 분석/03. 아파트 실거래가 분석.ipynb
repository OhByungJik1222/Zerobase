```python
# 부산, 서울 데이터 나눠서 저장하기
import os

def split_by_city(df, file_name) : 
    cities = np.unique(df['city'])
    for city in cities:
        if city not in os.listdir('data'): 
            os.mkdir(f'.\\data\\{city}')
        df[df['city'] == city].to_csv(f'.\\data\\{city}\\{file_name}.csv')
```

```python
location_df = pd.read_csv('.\\동별좌표.csv')
```

```python
split_by_city(location_df, "location_df")
```

```python
split_by_city(apartment_df, 'apartment_df')
split_by_city(day_care_center_df, 'day_care_center_df')
split_by_city(park_df, 'park_df')
```
