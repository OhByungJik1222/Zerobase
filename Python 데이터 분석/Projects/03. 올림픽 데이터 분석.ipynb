## **Data**

```python
import pandas as pd

df_target = pd.read_csv('.\\datas\\Summer-Olympic-medals-1976-to-2008.csv', encoding='cp1252')
df_target.head()
```

```python
df_target.info()
```

```python
# 결측치 처리
df_target = df_target.dropna().reset_index(drop=True)
df_target.info()
```

```python
# Year 컬럼 Data Type 변환
df_target['Year'] = df_target['Year'].astype(int)
df_target.info()
```

```python
# 2008년 베이징 올림픽 양궁 종목 대한민국 금메달 리스트
df_target[df_target['Sport'] == 'Archery']['Discipline'].value_counts()
```

```python
tmp_df = df_target.copy()
df_archery = tmp_df[(tmp_df['City'] == 'Beijing') & (tmp_df['Year'] == 2008)]
df_archery = df_archery[(df_archery['Sport'] == 'Archery') & (df_archery['Medal'] == 'Gold')]
df_archery = df_archery[df_archery['Country_Code'] == 'KOR']
df_archery
```

```python
# 대한민국 역대 (1976 ~ 2008) 하계 올림픽 메달 현황
tmp_df = df_target.copy()
display(tmp_df.info())

df_kor = tmp_df[tmp_df['Country_Code'] == 'KOR']
display(df_kor.info())

check_dup_cols = [col for col in df_kor.columns if col not in ['Athlete', 'Gender']]
df_kor = df_kor.drop_duplicates(subset=check_dup_cols)
display(df_kor.info())
```

```python
# Year과 Medal 컬럼으로 그루핑
grouped_df_kor = df_kor.groupby(['Year', 'Medal'])['City'].count().reset_index(name='City')
grouped_df_kor
```

```python
# 멀티 인덱스 테이블 생성
grouped_df_kor = grouped_df_kor.set_index(['Year', 'Medal'])
grouped_df_kor
```

```python
# 정렬 (금, 은, 동)
sort_order = {'Gold': 0, 'Silver': 1, 'Bronze': 2}
sorted_df_kor = grouped_df_kor.sort_index(
    level=['Year', 'Medal'],
    ascending=[True, True],
    key=lambda x: x.map(sort_order) if x.name == 'Medal' else x
)
sorted_df_kor
```

```python
# 1996년 애틀랜타 올림픽 총 메달 개수 기준 상위 10개 국가
tmp_df = df_target.copy()
display(tmp_df.info())

tmp_df = tmp_df[tmp_df['City'] == 'Atlanta']

check_dup_cols = [col for col in tmp_df.columns if col not in ['Athlete', 'Gender']]
tmp_df = tmp_df.drop_duplicates(subset=check_dup_cols)
display(tmp_df.info())
```

```python
# 국가별 그루핑
grouped_df = tmp_df.groupby(['Country']).size().reset_index(name='Medal')
grouped_df.head()
```

```python
# 정렬 후 상위 10개 국가 추출
df_rank_10 = grouped_df.sort_values(by='Medal', ascending=False).head(10)
df_rank_10
```

```python
# 1996년 애틀랜타 올림픽 금, 은, 동메달 개수 기준 상위 10개 국가
tmp_df = df_target.copy()
df_1996 = tmp_df[tmp_df['Year'] == 1996]

check_dup_cols = [col for col in df_1996.columns if col not in ['Athlete', 'Gender']]
df_1996 = df_1996.drop_duplicates(subset=check_dup_cols)
df_1996.info()
```

```python
# 금, 은, 동메달 one-hot encoding
df_1996['Gold'] = (df_1996['Medal'] == 'Gold').astype(int)
df_1996['Silver'] = (df_1996['Medal'] == 'Silver').astype(int)
df_1996['Bronze'] = (df_1996['Medal'] == 'Bronze').astype(int)
df_1996.head()
```

```python
# 국가별 그루핑
grouped_df = df_1996.groupby(['Country'], as_index=False)[['Gold', 'Silver', 'Bronze']].sum()
grouped_df.head()
```

```python
# 정렬 후 상위 10개 추출
df_rank_10 = grouped_df.sort_values(by=['Gold', 'Silver', 'Bronze'],
                                    ascending=[False, False, False]).head(10)
df_rank_10
```
