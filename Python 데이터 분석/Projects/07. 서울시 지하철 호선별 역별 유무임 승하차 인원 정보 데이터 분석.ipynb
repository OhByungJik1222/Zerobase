```python
# Raw Data Loading
import pandas as pd

df_target = pd.read_csv('./서울시 지하철 호선별 역별 유_무임 승하차 인원 정보.csv', encoding='cp949')
df_target.head()
```

```python
df_target.info()
```

```python
# 중복 데이터 제거
tmp_df = df_target.copy()
tmp_df = tmp_df.drop_duplicates(keep='first')

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target.info()
```

```python
# 사용월, 호선명, 지하철역의 데이터가 중복되는 행에 대해서
# 유임승차인원, 무임승차인원, 유임하차인원, 무임하차인원에 각각 합한 값을 저장하고
# 첫번째 데이터만 남기기
drop_columns = ['사용월','호선명','지하철역']
sum_columns = ['유임승차인원', '무임승차인원', '유임하차인원', '무임하차인원']

tmp_df = df_target.copy()

tmp_df[sum_columns] = tmp_df.groupby(drop_columns)[sum_columns].transform('sum')

tmp_df = tmp_df.drop_duplicates(subset=drop_columns, keep='first')

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target['호선명'].unique()
```

```python
# 호선명이 9호선2~3단계 또는 9호선2단계인 데이터를 9호선으로 변경
# 지하철역 컬럼의 데이터에 괄호 '(', ')'가 있다면 괄호와 괄호 안의 내용 삭제
# 지하철역 컬럼의 데이터가 신천이라면 잠실새내로 변경 
tmp_df = df_target.copy()

tmp_df['호선명'] = tmp_df['호선명'].replace(['9호선2~3단계', '9호선2단계'], '9호선')
tmp_df['지하철역'] = tmp_df['지하철역'].str.split('(').str[0].str.strip()
tmp_df['지하철역'] = tmp_df['지하철역'].replace('신천', '잠실새내')

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target['호선명'].unique()
```

```python
# incheon_station_list는 모두 7호선의 역이며 서울특별시가 관리하는 역이 아님
# 따라서 리스트에 있는 역들의 호선명 데이터를 '7호선'에서 '7호선 인천'으로 변경
incheon_station_list = ['까치울', '부천종합운동장', '춘의', '신중동', '부천시청', '상동', '삼산체육관', '굴포천', '부평구청', '산곡', '석남']

tmp_df = df_target.copy()

tmp_df.loc[(tmp_df['호선명'] == '7호선') &
           (tmp_df['지하철역'].isin(incheon_station_list)),
           '호선명'] = '7호선 인천'

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target['호선명'].unique()
```

```python
# 2022년 6월 기준 각 지하철역의
# 유임승차인원, 무임승차인원, 유임하차인원, 무임하차인원 합산
columns = ['유임승차인원', '무임승차인원', '유임하차인원', '무임하차인원']

tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['사용월'] == 202206]
tmp_df = tmp_df.groupby('지하철역')[columns].sum()

df_result = tmp_df.copy()
df_result.head()
```

```python
# 2022년 6월 기준 각 지하철역의 호선의 수가 3개 이상인 역
tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['사용월'] == 202206]
tmp_df = tmp_df.groupby(['지하철역'])['호선명'].nunique()
tmp_df = tmp_df[tmp_df >= 3].reset_index()

tmp_df.columns = ['지하철역', '호선수']
tmp_df = tmp_df.set_index('지하철역')
tmp_df = tmp_df.sort_values(by=['호선수', '지하철역'], ascending=[False, False])

df_result = tmp_df.copy()
df_result.head()
```

```python
# 위에서 구한 호선수가 3개 이상인 역과 해당 역의 환승 가능 호선 구하기
tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['사용월'] == 202206]

line_df = tmp_df.groupby('지하철역')['호선명'].nunique()
transfer_df = tmp_df.groupby('지하철역')['호선명'].apply(lambda x: ', '.join(sorted(x.unique())))

tmp_df = pd.concat([line_df, transfer_df], axis=1)
tmp_df.columns = ['호선수', '환승호선']

tmp_df = tmp_df[tmp_df['호선수'] >= 3]
tmp_df = tmp_df.sort_values(by=['호선수', '환승호선'], ascending=[False, True])

df_result = tmp_df.copy()
df_result.head()
```

```python
# 호선수가 3개 이상인 지하철역들의 유임승하차율, 무임승하차율, 승하차율 컬럼 추가하기
# 그 중 무임승하차율이 승하차율보다 크고, 102 이상인 역 출력
import warnings

warnings.filterwarnings('ignore')

def get_ratio(df, ratio_type):
    if ratio_type == '유임승하차율':
        return round(df['유임승차인원'].sum() / df['유임하차인원'].sum() * 100, 2)
    elif ratio_type == '무임승하차율':
        return round(df['무임승차인원'].sum() / df['무임하차인원'].sum() * 100, 2)
    elif ratio_type == '승하차율':
        return round((df['유임승차인원'].sum() + df['무임승차인원'].sum()) / 
                     (df['유임하차인원'].sum() + df['무임하차인원'].sum()) * 100, 2)


ratio_columns = ['유임승하차율', '무임승하차율', '승하차율']
columns = ['유임승차인원', '무임승차인원', '유임하차인원', '무임하차인원']

tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['사용월'] == 202206]

line_df = tmp_df.groupby('지하철역')['호선명'].nunique()
line_df = line_df[line_df >= 2]

ratio_df = pd.DataFrame(index=line_df.index)
for col in ratio_columns:
    ratio_df[col] = tmp_df.groupby('지하철역')[columns].apply(get_ratio, col)

ratio_df['호선수'] = line_df

ratio_df = ratio_df[(ratio_df['무임승하차율'] > ratio_df['승하차율']) &
                    (ratio_df['무임승하차율'] >= 102)]

ratio_df = ratio_df[['호선수'] + ratio_columns]
ratio_df = ratio_df.sort_values(by=['호선수', '무임승하차율'], ascending=[False, False])

df_result = ratio_df.copy()
df_result
```

```python
# 2022년 1월 대비 2022년 6월에 유동인구 증감율 구하기
# 이 중 유동인구가 감소한(유동인구 증감율이 0보다 작은) 역 출력
columns = ['유임승차인원', '무임승차인원', '유임하차인원', '무임하차인원']

tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['호선명'] != '7호선 인천']

jan_df = tmp_df[tmp_df['사용월'] == 202201]
jun_df = tmp_df[tmp_df['사용월'] == 202206]

jan_df['유동인구'] = jan_df[columns].sum(axis=1)
jun_df['유동인구'] = jun_df[columns].sum(axis=1)

jan_df = jan_df.set_index('지하철역')
jun_df = jun_df.set_index('지하철역')

jan_df = jan_df.groupby('지하철역')['유동인구'].sum()
jun_df = jun_df.groupby('지하철역')['유동인구'].sum()

tmp_df = pd.DataFrame({
    '유동인구_1월': jan_df,
    '유동인구_6월': jun_df
})

tmp_df = tmp_df.dropna()

tmp_df['유동인구증감률'] = round((tmp_df['유동인구_6월'] / tmp_df['유동인구_1월'] - 1) * 100, 2)
tmp_df = tmp_df[tmp_df['유동인구증감률'] < 0][['유동인구증감률']]
tmp_df = tmp_df.sort_values(by='유동인구증감률', ascending=False)

df_result = tmp_df.copy()
df_result
```

```python
# 2015년에서 2022년까지 지하철 1~9호선에서 신설된 역 확인
# 해당 기간 중 유동인구 데이터가 존재하지 않는 경우 해당 기간에는 해당 역은 존재하지 않는다고 간주
columns = ['유임승차인원', '무임승차인원', '유임하차인원', '무임하차인원']
lines = [f'{i}호선' for i in range(1, 10)]

tmp_df = df_target.copy()

tmp_df = tmp_df[tmp_df['호선명'].isin(lines)]
tmp_df = tmp_df[(tmp_df['사용월'] >= 201501) &
                (tmp_df['사용월'] <= 202206)]

tmp_df['유동인구'] = tmp_df[columns].sum(axis=1)
tmp_df = tmp_df[tmp_df['유동인구'] > 0]
tmp_df = tmp_df.groupby(['호선명', '지하철역'])['사용월'].min()

tmp_df = tmp_df.reset_index()
tmp_df.columns = ['호선명', '지하철역', '신설월']
tmp_df = tmp_df.loc[tmp_df.groupby('지하철역')['신설월'].idxmin()]
tmp_df = tmp_df[tmp_df['신설월'] != 201501]

tmp_df = tmp_df.sort_values(by=['호선명', '신설월', '지하철역'], ascending=[True, False, True]).reset_index(drop=True)

df_result = tmp_df.copy()
df_result
```
