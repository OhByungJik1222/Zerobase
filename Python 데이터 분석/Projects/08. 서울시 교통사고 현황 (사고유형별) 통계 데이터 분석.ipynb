```python
# Raw Data Loading
import pandas as pd

df_target = pd.read_csv('./교통사고+현황(사고유형별).csv')
df_target.head()
```

```python
df_target.info()
```

```python
# 컬럼 정리
# '자치구별(1)', '사고유형별(2)' 컬럼 삭제
# '자치구별(2)' -> '자치구', '사고유형별(1)' -> '사고유형', '구분별(1)' -> '구분'
# 으로 컬럼명 수정
drop_columns = ['자치구별(1)', '사고유형별(2)']
rename_columns = {
    '자치구별(2)': '자치구',
    '사고유형별(1)': '사고유형',
    '구분별(1)': '구분'
}

tmp_df = df_target.copy()

tmp_df = tmp_df.drop(columns=drop_columns)
tmp_df = tmp_df.rename(columns=rename_columns)

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target.info()
```

```python
# '구분' 컬럼의 데이터 중 괄호와 괄호 안의 내용 삭제
tmp_df = df_target.copy()

tmp_df['구분'] = tmp_df['구분'].str.split('(').str[0]
tmp_df['구분'] = tmp_df['구분'].str.strip()

df_target = tmp_df.copy()
df_target.head()
```

```python
df_target.columns
```

```python
# '1988' ~ '2021' 컬럼의 데이터 중 결측치는 '-'으로 저장되어 있으므로 0으로 수정
# 이후 현재 object 타입인 '1988' ~ '2021' 컬럼들의 타입을 int 타입으로 변경
columns = [str(i) for i in range(1988, 2022)]

tmp_df = df_target.copy()

tmp_df[columns] = tmp_df[columns].replace('-', 0)
tmp_df[columns] = tmp_df[columns].fillna(0)
tmp_df[columns] = tmp_df[columns].astype(int)

df_target = tmp_df.copy()
df_target.info()
```

```python
# '자치구' 컬럼의 데이터 중 '소계' 데이터를 '서울시'로 변경
# '자치구', '사고유형', '구분' 별로 groupby
groupby_columns = ['자치구', '사고유형', '구분']

tmp_df = df_target.copy()
tmp_df.loc[tmp_df['자치구'] == '소계', '자치구'] = '서울시'

tmp_df = tmp_df.groupby(groupby_columns, sort=False).sum()

df_target = tmp_df.copy()
df_target.head()
```

```python
# 값이 0인 '1988' ~ '2004' 컬럼의 '자치구'-'합계'-'발생건수' 행의 데이터 채우기
# 서울시 포함 각 자치구의 '합계'-'발생건수'는
# ('차대사람'-'발생건수') + ('차대차'-'발생건수') + ('차량단독'-'발생건수') + ('건널목'-'발생건수')
# 이때 '차대사람', '차대차', '차량단독', '건널목' 행이 없는 경우가 있으므로 주의
columns = [str(i) for i in range(1988, 2005)]

tmp_df = df_target.copy()

for gu in tmp_df.index.get_level_values(0).unique():
    target_idx = (gu, '합계', '발생건수')

    idx_list = [
        (gu, '차대사람', '발생건수'),
        (gu, '차대차', '발생건수'),
        (gu, '차량단독', '발생건수'),
        (gu, '건널목', '발생건수')
    ]

    exist_idx = [idx for idx in target_idx if idx in tmp_df.index]

    sum_value = tmp_df.loc[exist_idx, columns].sum()

    is_zero = tmp_df.loc[target_idx, columns] == 0
    tmp_df.loc[target_idx, columns] = tmp_df.loc[target_idx, columns].mask(is_zero, sum_value)

df_target = tmp_df.copy()
df_target.head()
```

```python
# '구분' 인덱스 삭제 (발생건수만 추출)
df_result = df_target.xs('발생건수', level='구분')

df_result
```

```python
# 서울시 포함 각 자치구의 사망자수 집계
type_list = ['합계', '차대사람', '차대차', '차량단독', '건널목']
gu_list = ['서울시', '종로구', '중구', '용산구', '성동구', '광진구', '동대문구', '중랑구', '성북구', '강북구', '도봉구', '노원구', '은평구', '서대문구', '마포구', '양천구', '강서구', '구로구', '금천구', '영등포구', '동작구', '관악구', '서초구', '강남구', '송파구', '강동구']
columns = [str(i) for i in range(2005, 2022)]

df_result = df_target.xs('사망자수', level='구분')
df_result = df_result[columns]

df_result = df_result.reorder_levels(['사고유형', '자치구'])

df_result.index = df_result.index.set_levels([
    pd.CategoricalIndex(df_result.index.levels[0], categories=type_list, ordered=True),
    pd.CategoricalIndex(df_result.index.levels[1], categories=gu_list, ordered=True)
])
df_result = df_result.sort_index(level=['사고유형','자치구'])

df_result.head()
```

```python
# 2005년 이후 서울시 포함 각 자치구 사망률 집계
type_list = ['합계', '차대사람', '차대차', '차량단독', '건널목']
gu_list = ['서울시', '종로구', '중구', '용산구', '성동구', '광진구', '동대문구', '중랑구', '성북구', '강북구', '도봉구', '노원구', '은평구', '서대문구', '마포구', '양천구', '강서구', '구로구', '금천구', '영등포구', '동작구', '관악구', '서초구', '강남구', '송파구', '강동구']
columns = [str(i) for i in range(2005, 2022)]

deaths = df_target.xs('사망자수', level='구분')[columns]
accidents = df_target.xs('발생건수', level='구분')[columns]
df_result = (deaths / accidents * 100).round(2)

df_result.index = df_result.index.set_levels([
    pd.CategoricalIndex(df_result.index.levels[0], categories=gu_list, ordered=True),
    pd.CategoricalIndex(df_result.index.levels[1], categories=type_list, ordered=True)
])
df_result = df_result.sort_index(level=['자치구', '사고유형'])
df_result = df_result.dropna()

df_result.head(10)
```
