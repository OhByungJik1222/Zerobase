```python
import pandas as pd

# 서울시 구별/사고유형별 교통사고 현황
path = './교통사고+현황(사고유형별)_preprocessing.csv'
df_acc = pd.read_csv(path)
print('서울시 구별/사고유형별 교통사고 현황')
display(df_acc.head())

# 서울시 구별 차량통행속도
path = './차량통행속도(월별_구별).csv'
df_speed = pd.read_csv(path)
print('서울시 구별 차량통행속도')
display(df_speed.head())


# 서울시 구별 인구밀도(단위: 인구(명) / 면적(㎢) / 인구밀도(명/㎢))
path = './인구밀도.csv'
df_density = pd.read_csv(path)
print('서울시 구별 인구밀도')
display(df_density.head())
```

```python
# df_speed 컬럼 수정
# '구분별(1)' 컬럼 삭제, '구분별(2)' -> '자치구'
# 또한 '자치구' 컬럼의 데이터 중 '강남', '강북' 데이터 제외
# '자치구' 컬럼의 데이터 중 '소계' 데이터 '서울시'로 변경
# '구분' 컬럼 추가 후 해당 컬럼에 '차량통행속도' 데이터 저장
ordered_columns = ['자치구', '구분'] + [str(i) for i in range(2013, 2022)]

tmp_df = df_speed.copy()

tmp_df = tmp_df[list(tmp_df.columns[1:])]
tmp_df = tmp_df.rename(columns={tmp_df.columns[0]: '자치구'})

tmp_df.loc[tmp_df['자치구'] == '소계', '자치구'] = '서울시'
tmp_df = tmp_df[~tmp_df['자치구'].isin(['강남', '강북'])]

tmp_df['구분'] = '차량통행속도'
tmp_df = tmp_df[ordered_columns]
tmp_df = tmp_df.reset_index(drop=True)

df_speed = tmp_df.copy()
df_speed.head()
```

```python
# df_density 컬럼 및 데이터 수정
# '동별(2)' 컬럼명 '자치구'로 수정
# '자치구' 컬럼의 데이터 중 '소계' 데이터 '서울시'로 수정
# '동별(1)' 컬럼 삭제
# 컬럼 이름에 '.1', '.2'가 포함되어 있다면 삭제 (ex. '2022.2' -> '2022')
# '구분' 컬럼 생성 후 해당 컬럼에 '인구', '면적', '인구밀도' 데이터 저장
# '구분' 컬럼의 데이터에 맞게 '2013' ~ '2021' 컬럼 데이터 채우기 
columns = ['자치구', '구분'] + [str(i) for i in range(2013, 2022)]

tmp_df = df_density.copy()

tmp_df = tmp_df[tmp_df.columns[1:]]
tmp_df = tmp_df.rename(columns={tmp_df.columns[0]: '자치구'})
tmp_df.loc[tmp_df['자치구'] == '소계', '자치구'] = '서울시'
tmp_df = tmp_df.drop(index=0)
tmp_df.columns = tmp_df.columns.str.replace(r'\.1|\.2', '', regex=True)

tmp_df = tmp_df.melt(id_vars=['자치구'], var_name='년도', value_name='값')
tmp_df['구분'] = tmp_df.groupby(['자치구', '년도']).cumcount().map({0:'인구', 1:'면적', 2:'인구밀도'})
tmp_df = tmp_df.pivot_table(
    index=['자치구', '구분'],
    columns='년도',
    values='값',
    aggfunc='first',
    sort=False
).reset_index()

tmp_df.columns.name = None
tmp_df = tmp_df[columns]
for col in columns[2:]:
    tmp_df[col] = tmp_df[col].astype(float)

df_density = tmp_df.copy()
df_density.head()
```

```python
# 처리한 df_acc, df_speed, df_density 합치기
gu_list = ['서울시', '종로구', '중구', '용산구', '성동구', '광진구', '동대문구', '중랑구', '성북구', '강북구', '도봉구', '노원구', '은평구', '서대문구', '마포구', '양천구', '강서구', '구로구', '금천구', '영등포구', '동작구', '관악구', '서초구', '강남구', '송파구', '강동구']
cat_list = ['발생건수', '사망자수', '부상자수', '차량통행속도', '인구', '면적', '인구밀도']
type_list = ['합계', '차대사람', '차대차', '차량단독', '건널목', '-']

tmp_df = pd.concat([df_acc, df_speed, df_density], axis=0, ignore_index=True)
tmp_df.loc[tmp_df['구분'].isin(['차량통행속도', '인구', '면적', '인구밀도']) &
           tmp_df['사고유형'].isna(), '사고유형'] = '-'

tmp_df['자치구'] = pd.Categorical(tmp_df['자치구'], categories=gu_list, ordered=True)
tmp_df['구분'] = pd.Categorical(tmp_df['구분'], categories=cat_list, ordered=True)
tmp_df['사고유형'] = pd.Categorical(tmp_df['사고유형'], categories=type_list, ordered=True)
tmp_df = tmp_df.sort_values(['자치구', '사고유형', '구분']).reset_index(drop=True)

df_target = tmp_df.copy()
df_target.head(20)
```

```python
# 종로구의 cat_list 구하기 
cat_list = ['인구 만명당 발생건수', '인구 만명당 사망자수', '인구 만명당 부상자수',
            '인구밀도당 부상자수', '인구밀도당 사망자수', '인구밀도당 발생건수',
            '차량통행속도당 사망자수', '차량통행속도당 발생건수', '차량통행속도당 부상자수']
type_list = ['합계', '차대사람', '차대차', '차량단독']
columns = [str(i) for i in range(2013, 2022)]
target_columns = ['발생건수', '사망자수', '부상자수']

jongno_df = df_target[df_target['자치구'] == '종로구'].copy()

population = jongno_df[jongno_df['구분'] == '인구'][columns].values[0]
density = jongno_df[jongno_df['구분'] == '인구밀도'][columns].values[0]
speed = jongno_df[jongno_df['구분'] == '차량통행속도'][columns].values[0]

population_df = jongno_df[jongno_df['구분'].isin(target_columns)].copy()
population_df['구분'] = '인구 만명당 ' + population_df['구분'].astype(str)
population_df[columns] = population_df[columns].div(population / 10000, axis=1).round(2)

density_df = jongno_df[jongno_df['구분'].isin(target_columns)].copy()
density_df['구분'] = '인구밀도당 ' + density_df['구분'].astype(str)
density_df[columns] = (density_df[columns].values / density).round(2)

speed_df = jongno_df[jongno_df['구분'].isin(target_columns)].copy()
speed_df['구분'] = '차량통행속도당 ' + speed_df['구분'].astype(str)
speed_df[columns] = (speed_df[columns].values / speed).round(2)

df_result = pd.concat([population_df, density_df, speed_df], axis=0)

df_result['사고유형'] = pd.Categorical(df_result['사고유형'], categories=type_list, ordered=True)
df_result['구분'] = pd.Categorical(df_result['구분'], categories=cat_list, ordered=True)
df_result = df_result.sort_values(['사고유형', '구분']).reset_index(drop=True)

df_result.head(10)
```

```python
# 서울시 포함 전체 자치구의 cat_list 구하기
gu_list = ['서울시', '종로구', '중구', '용산구', '성동구', '광진구', '동대문구', '중랑구', '성북구', '강북구', '도봉구', '노원구', '은평구', '서대문구', '마포구', '양천구', '강서구', '구로구', '금천구', '영등포구', '동작구', '관악구', '서초구', '강남구', '송파구', '강동구']
cat_list = ['인구 만명당 발생건수', '인구 만명당 사망자수', '인구 만명당 부상자수', '인구밀도당 발생건수', '인구밀도당 사망자수', '인구밀도당 부상자수',  '차량통행속도당 발생건수', '차량통행속도당 사망자수',  '차량통행속도당 부상자수']
type_list = ['합계', '차대사람', '차대차', '차량단독', '건널목']
columns = [str(i) for i in range(2013, 2022)]
target_columns = ['발생건수', '사망자수', '부상자수']

result_list = []
for gu in gu_list:
    tmp_df = df_target[df_target['자치구'] == gu].copy()
    
    population = tmp_df[tmp_df['구분']=='인구'][columns].values[0]
    density = tmp_df[tmp_df['구분']=='인구밀도'][columns].values[0]
    speed = tmp_df[tmp_df['구분']=='차량통행속도'][columns].values[0]
    
    population_df = tmp_df[tmp_df['구분'].isin(target_columns)].copy()
    population_df['구분'] = '인구 만명당 ' + population_df['구분'].astype(str)
    population_df[columns] = population_df[columns].div(population/10000, axis=1).round(2)
    result_list.append(population_df)
    
    density_df = tmp_df[tmp_df['구분'].isin(target_columns)].copy()
    density_df['구분'] = '인구밀도당 ' + density_df['구분'].astype(str)
    density_df[columns] = (density_df[columns].values / density).round(2)
    result_list.append(density_df)
    
    speed_df = tmp_df[tmp_df['구분'].isin(target_columns)].copy()
    speed_df['구분'] = '차량통행속도당 ' + speed_df['구분'].astype(str)
    speed_df[columns] = (speed_df[columns].values / speed).round(2)
    result_list.append(speed_df)

df_result = pd.concat(result_list, axis=0)

df_result['자치구'] = pd.Categorical(df_result['자치구'], categories=gu_list, ordered=True)
df_result['사고유형'] = pd.Categorical(df_result['사고유형'], categories=type_list, ordered=True)
df_result['구분'] = pd.Categorical(df_result['구분'], categories=cat_list, ordered=True)
df_result = df_result.sort_values(['자치구','사고유형','구분']).reset_index(drop=True)

df_result.head(10)
```

```python
# '사고유형' 컬럼이 '합계'인 데이터 중 각 '구분'별 및 '연도'별 
# 값이 최대인 서울시 포함 자치구 구하기
columns = ['사고유형', '구분', '연도', '자치구', '값']
cat_list = ['인구 만명당 발생건수', '인구밀도당 발생건수', '차량통행속도당 발생건수']
years = [str(i) for i in range(2013, 2022)]

total_df = df_result[(df_result['사고유형'] == '합계') &
                     (df_result['자치구'] != '서울시')]
total_df = total_df[total_df['구분'].isin(cat_list)]

total_df = total_df.melt(id_vars=['사고유형', '자치구', '구분'],
                         value_vars=years,
                         var_name='연도',
                         value_name='값')

idx = total_df.groupby(['구분', '연도'], observed=True)['값'].idxmax()

total_df = total_df.loc[idx, columns]

total_df = total_df.sort_values(['구분', '연도']).reset_index(drop=True)

df_answer = total_df.copy()
df_answer
```
